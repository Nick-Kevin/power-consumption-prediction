This is a regression task that is designed to forecast the global active power  of a household.  
The dataset used to train the models is the [Individual Household Electric Power Consumption](https://archive.ics.uci.edu/dataset/235/individual+household+electric+power+consumption) from UCI Machine Learning Repository

```python
import pandas as pd
import numpy as np
import pickle
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, ElasticNet # Ridge instead of Lasso because most of features are relevant for predictions
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
data = pd.read_csv('household_power_consumption.txt', sep=';', low_memory=False)

print(data.head())
```

# Data preprocessing + EDA

```python
print(data.info(show_counts=True))
```

```python
print((data['Global_active_power'] == '?').sum())
```

```python
cols = data.columns.tolist()
print(cols)
```

```python
for col in cols:
    empty_count = (data[col] == '?').sum()
    print('Missing values in ' + col + ' = ' + str(empty_count))
```

```python
# converting "?" into np.nan
cols_containing_missing = ['Global_active_power', 'Global_reactive_power', 'Voltage', 'Global_intensity', 'Sub_metering_1', 'Sub_metering_2']

data[cols_containing_missing] = data[cols_containing_missing].replace('?', np.nan)
print(data.info(show_counts=True))
```

```python
# remove missing values
data.dropna(inplace=True)

print(data.info(show_counts=True))
```

```python
# convert str features into float, exept Date and Time
for col in cols_containing_missing:
    data[col] = pd.to_numeric(data[col])

print(data.info(show_counts=True))
```

```python
data.describe()
```

Separate the target variable from the independant variables

```python
X = data.drop(columns=['Global_active_power'])
target = data['Global_active_power']

X.head()
```

Processing date and time

```python
# combine Date and Time into only one features and drop them
X['Date_time'] = X['Date'] + ' ' + X['Time']
X.drop(columns=['Date', 'Time'], inplace=True)
X['Date_time'] = pd.to_datetime(X['Date_time'], dayfirst=True)
```

```python
X['Date_time'].head(2)
```

```python
# extract necessary informations form date
X['month'] = X['Date_time'].dt.month
X['day'] = X['Date_time'].dt.day
X['year'] = X['Date_time'].dt.year
X['Day_of_week'] = X['Date_time'].dt.dayofweek
X['Week_of_year'] = X['Date_time'].dt.isocalendar().week

X.head(2)
```

```python
# extract necessary informations from the time feature

X['Hour'] = X['Date_time'].dt.hour
X['Minute'] = X['Date_time'].dt.minute

X.head(2)
```

```python
# drop the date-Time feature

X.drop(columns=['Date_time'], inplace=True)

X.head(1)
```

```python
# drawing boxplot to visualize outliers
sensor_cols = ['Global_reactive_power', 'Voltage', 'Global_intensity', 'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']

plt.figure(figsize=(16, 12))

for col in sensor_cols:
    plt.subplot(3, 3, sensor_cols.index(col)+1)
    sns.boxplot(data=X[col])
    plt.title('Boxplot of ' + col)

plt.show()
```

```python
# compute the correlation matrix, then plot
X_corr = X.corr()

plt.figure(figsize=(12, 12))
sns.heatmap(X_corr, annot=True)
plt.title('Correlation matrix | Relationship between features')
plt.show()
```

```python
# drop month column because this information is in week of year
X.drop(columns='month', inplace=True)
X.info(show_counts=True)
```

Data split

```python
X_train, X_test, target_train, target_test = train_test_split(X, target, test_size=0.3, random_state=42)
```

Scale each features by using Standardization because, on the one hand, algorithms which assume normally distributed data like Linear regression and SVM are going to be used, on the other hand, some features contain outliers

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# store the scaler in a pickle file
with open("scaler.pkl", 'wb') as file:
    pickle.dump(scaler, file)
```

# Model training, test and evaluation

```python
# model evaluation function
def evaluate (model, target_test, target_pred):
    """
        display the evaluation metrics for a model

        Arguments:
            model: (string) the name of the model
            target_test: (numpy.ndarray) the test set for the target feature
            target_pred: (numpy.ndarray) the predicted values of the model
    """

    assert type(model) == str
    assert target_test.shape == target_pred.shape

    mse = mean_squared_error(target_test, target_pred)
    mae = mean_absolute_error(target_test, target_pred)
    r2 = r2_score(target_test, target_pred)
    print('---------------- ' + model + ' evaluation ----------------')
    print('Mean Squared Error (MSE) for ' + model + ' = ' + str(mse))
    print('Mean Absolute Error (MAE) for ' + model + ' = ' + str(mae))
    print('Coefficient of determination (R2) for ' + model + ' = ' + str(r2))
```

```python
lr_model = LinearRegression()
ridge_model = Ridge(random_state=42)
elastic_net_model = ElasticNet(random_state=42) # Lasso (L1 regularization) + Ridge (L2)
dt_model = DecisionTreeRegressor(random_state=42)
rf_model = RandomForestRegressor(
    n_estimators=50,      # Reduce number of trees from 100 to 50
    max_depth=12,         # Strictly limit how deep the trees can grow
    min_samples_leaf=20,  # Prevents creating millions of tiny, memory-heavy leaves
    max_samples=0.2,      # Train each tree on only 20% of the data (cuts RAM usage significantly)
    n_jobs=-1,
    random_state=42
)
#lr_svr = LinearSVR(max_iter=10000)
```

```python
lr_model.fit(X_train, target_train)
```

```python
ridge_model.fit(X_train, target_train)
```

```python
elastic_net_model.fit(X_train, target_train)
```

```python
dt_model.fit(X_train, target_train)
```

```python
rf_model.fit(X_train, target_train)
```

```python
lr_predict = lr_model.predict(X_test)
evaluate('Linear Regression', target_test, lr_predict)
```

```python
ridge_pred = ridge_model.predict(X_test)
evaluate('Ridge', target_test, ridge_pred)
```

```python
elastic_net_pred = elastic_net_model.predict(X_test)
evaluate('Elastic Net', target_test, elastic_net_pred)
```

```python
dt_pred = dt_model.predict(X_test)
evaluate('Decision Tree', target_test, dt_pred)
```

```python
rf_pred = rf_model.predict(X_test)
evaluate('Random', target_test, rf_pred)
```

```python
# export the rf model
with open("rf-model.pkl", "wb") as file:
    pickle.dump(rf_model, file)
```
